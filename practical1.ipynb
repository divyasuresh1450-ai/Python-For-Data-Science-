```python


from array import array

print("STRING OPERATIONS")
s = "Python"
print("Original String:", s)
print("Length:", len(s))
print("First character:", s[0])
print("Last character:", s[-1])
print("Substring:", s[0:3])
print("Uppercase:", s.upper())
print("Concatenation:", s + " Programming")
print("Repetition:", s * 2)

print("\n LIST OPERATIONS ")
lst = [10, 20, 30, 40]
print("Original List:", lst)
lst.append(50)
print("After append:", lst)
lst.insert(1, 15)
print("After insert:", lst)
lst.remove(30)
print("After remove:", lst)
print("Slicing:", lst[1:4])
print("Length:", len(lst))
lst.sort()
print("Sorted List:", lst)

print("\n ARRAY OPERATIONS")
arr = array('i', [1, 2, 3, 4, 5])
print("Original Array:", arr)
arr.append(6)
print("After append:", arr)
arr.insert(2, 10)
print("After insert:", arr)
arr.remove(3)
print("After remove:", arr)
print("Access element:", arr[0])

print("\n TUPLE OPERATIONS")
tup = (100, 200, 300, 400)
print("Original Tuple:", tup)
print("First element:", tup[0])
print("Slicing:", tup[1:3])
print("Length:", len(tup))
print("Count of 200:", tup.count(200))
print("Index of 300:", tup.index(300))

print("\n DICTIONARY OPERATIONS")
d = {"name": "Divya", "age": 21, "city": "Nagpur"}
print("Original Dictionary:", d)
print("Name:", d["name"])
d["age"] = 22
print("Updated Dictionary:", d)
d["course"] = "Python"
print("After adding new key:", d)
print("Keys:", d.keys())
print("Values:", d.values())

print("\n SET OPERATIONS")
st = {1, 2, 3, 4}
print("Original Set:", st)
st.add(5)
print("After add:", st)
st.remove(2)
print("After remove:", st)
print("Union:", st.union({6, 7}))
print("Intersection:", st.intersection({3, 4, 8}))

print("\n RANGE OPERATIONS ")
r = range(1, 10, 2)
print("Range values:")
for i in r:
    print(i, end=" ")
print("\nLength of range:", len(r))
```

1) What are sequence data types in Python?

Name them and explain how indexing and slicing work across different sequence types.

Sequence data types are data types that store multiple values in an ordered manner.
Each element has a position (index), which allows accessing specific items.

String → sequence of characters
List → ordered, mutable collection
Tuple → ordered, immutable collection
Range → sequence of numbers
Array → same-type elements (from array module)

Although dictionary and set are collection types, they are not true sequence types because they do not support normal positional indexing like lists or tuples.

Indexing

Indexing means accessing elements using their position.

Python uses zero-based indexing.

lst = [10, 20, 30, 40]
print(lst[0])   # 10
print(lst[2])   # 30

Negative indexing starts from the end.

print(lst[-1])  # 40
Slicing

Slicing means extracting a part of the sequence.

Syntax:

sequence[start:stop:step]

Example:

s = "Python"
print(s[1:4])

4) Explain how dictionaries work internally in Python.
Why are dictionary lookups faster compared to lists or tuples?

A dictionary stores data as key-value pairs.
Example:
d = {"name": "Divya", "age": 20}
Dictionaries use a hash table internally.
When a key is inserted:

Python applies a hash function to the key
Generates a unique hash value
Stores the value at that memory location

Example:

d["name"]

Python directly uses the hash of "name" to find the value.

Why faster?

Dictionary lookup is usually O(1) (constant time).

This means Python can directly access the value without checking every item.

Whereas lists and tuples require linear search O(n) if you are searching for a value.

Example:

# list search
if 30 in [10,20,30,40]:

This checks elements one by one.

That is why dictionaries are much faster for searching.
