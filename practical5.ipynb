```python
import matplotlib.pyplot as plt

plt.scatter(df['Age'], df['Daily_Usage_Hours'])
plt.xlabel("Age")
plt.ylabel("Daily Usage Hours")
plt.title("Age vs Daily Usage Hours")
plt.show()
```

```python
import matplotlib.pyplot as plt

df_sorted = df.sort_values('Age')

plt.plot(df_sorted['Age'], df_sorted['Daily_Usage_Hours'])
plt.title("Line Plot")
plt.xlabel("Age")
plt.ylabel("Daily Usage Hours")
plt.show()
```

```python
import matplotlib.pyplot as plt

avg_usage = df.groupby('Education_Level')['Daily_Usage_Hours'].mean()

plt.bar(avg_usage.index, avg_usage.values)
plt.xlabel("Education Level")
plt.ylabel("Average Daily Usage Hours")
plt.title("Bar Chart")
plt.show()
```

```python
plt.plot(df['Student_ID'], df['Daily_Usage_Hours'])
plt.title("Student ID vs Daily Usage Hours")
plt.xlabel("Student ID")
plt.ylabel("Daily Usage Hours")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.hist(df['Daily_Usage_Hours'], bins=10)
plt.title("Histogram")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Frequency")
plt.show()
```

1. What is the importance of data visualization in data analysis? How does visualization help in exploratory data analysis (EDA)?

Data visualization is the graphical representation of data using charts, graphs, and plots. It helps in understanding large and complex datasets easily.
Importance:Makes data easy to understand,Helps identify patterns and trends,Helps compare values
Detects outliers and anomalies,Supports decision-making
Role in EDA:In Exploratory Data Analysis (EDA), visualization helps to,understand data distribution,identify missing values,find relationships between variables.
Examples include histogram, scatter plot, box plot, and line graph.


2. What is the difference between Matplotlib and other visualization libraries?

Matplotlib is a basic Python library used for creating static, animated, and customized graphs.
Difference from other libraries:Seaborn: built on Matplotlib, easier and more visually attractive.
Plotly: provides interactive plots
Pandas plotting: simple graphs directly from datasets.
Matplotlib offers more customization and control, while other libraries focus on simplicity or interactivity.
