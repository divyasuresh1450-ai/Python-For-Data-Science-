1. Role of NumPy in Python Scientific Computing

NumPy is the core library for numerical computing in Python.Provides multidimensional arrays.Supports fast mathematical computations,
Acts as the base for libraries like Pandas, Matplotlib, Scikit-learn,Enables vectorized operations.

2. Why NumPy is Faster than Pure Python

Uses C-based implementation internally,Avoids Python loops uses vectorization,Efficient memory usage (continuous memory blocks),Supports broadcasting.

3. Mathematical Operations in NumPy


Arithmetic → + - * /,Statistical → mean, median, std,Linear algebra : dot product, matrix multiplication,Trigonometric : sin, cos,Logarithmic : log, exp..

4. Indexing and Slicing in NumPy
Indexing:

arr[0]       # first element
arr[-1]      # last element

Slicing:

arr[1:5]     # elements from index 1 to 4
arr[:, 1]    # all rows, column 1
