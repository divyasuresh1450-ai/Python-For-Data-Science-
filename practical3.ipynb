```python
import pandas as pd

df = pd.read_csv("aistudent.csv")
print(df.head())
```

```python
print(df.columns)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print(df.notnull().sum())
```

```python
df_clean = df.dropna()
print(df_clean)
```

```python
df["Daily_Usage_Hours"] = df["Daily_Usage_Hours"].fillna(df["Daily_Usage_Hours"].mean())
print(df)
```

```python
students_above_20 = df[df["Age"] > 20]
print(students_above_20)
```

```python
heavy_users = df[df["Daily_Usage_Hours"] > 3]
print(heavy_users)
```

```python
female_students = df[df["Gender"] == "Female"]
print(female_students)
```

```python
sorted_age = df.sort_values(by="Age")
print(sorted_age)
```

```python
print(df.groupby("Gender")["Daily_Usage_Hours"].mean())
```

```python
df["Usage_Category"] = df["Daily_Usage_Hours"].apply(
    lambda x: "High" if x > 3 else "Low"
)

print(df.head())
```

```python
df.rename(columns={"Daily_Usage_Hours": "Usage_Hours"}, inplace=True)
print(df.columns)
```

```python
df.drop("Student_ID", axis=1, inplace=True)
print(df.head())
```

```python
print(df.duplicated().sum())
```

```python
print(df["Satisfaction_Level"].value_counts())
```

1) What is data wrangling? Difference between data wrangling and data preprocessing.

Data wrangling is the process of cleaning, organizing, and transforming raw data into a useful format for analysis.
It includes:removing missing values,deleting duplicates,changing column names.

Data processing:Prepares data for ML models
Handles missing values and formatting	Includes scaling and encoding

2) Importance of EDA in Pandas

EDA (Exploratory Data Analysis) helps us understand the dataset before analysis.Check rows and columns,find missing values,identify duplicates
understand data distribution.

3) Difference between Series, DataFrame, and Panel

Series: 1-dimensional data
DataFrame: 2-dimensional table (rows and columns)
Panel: 3-dimensional data (multiple DataFrames)
