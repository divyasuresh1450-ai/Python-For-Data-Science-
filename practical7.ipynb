```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10      # number of trials
p = 0.5     # probability of success

x = np.arange(0, n+1)
prob = binom.pmf(x, n, p)

plt.bar(x, prob)
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.show()
```

```python
from scipy.stats import poisson

lam = 3   # average rate

x = np.arange(0, 10)
prob = poisson.pmf(x, lam)

plt.bar(x, prob)
plt.title("Poisson Distribution (λ=3)")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.show()
```

```python
from scipy.stats import norm

mu = 0       # mean
sigma = 1    # standard deviation

x = np.linspace(-5, 5, 100)
y = norm.pdf(x, mu, sigma)

plt.plot(x, y)
plt.title("Normal Distribution (μ=0, σ=1)")
plt.xlabel("Values")
plt.ylabel("Density")
plt.show()
```

1. What is the difference between Binomial, Poisson, and Normal distributions?


| Feature    | Binomial Distribution       | Poisson Distribution      | Normal Distribution         |
| ---------- | --------------------------- | ------------------------- | --------------------------- |
| Type       | Discrete                    | Discrete                  | Continuous                  |
| Used for   | Fixed number of trials      | Rare events over interval | Natural measurements        |
| Parameters | n (trials), p (probability) | λ (rate of occurrence)    | μ (mean), σ (std deviation) |
| Example    | Tossing a coin n times      | Number of calls per hour  | Heights of people           |
| Shape      | Can be symmetric/skewed     | Positively skewed         | Symmetric bell curve        |

Number of accidents at a traffic signal per day

Significance:

Applied in healthcare (disease outbreak tracking)


3. What is the Central Limit Theorem and how does it relate to Normal distribution?

The Central Limit Theorem states that:
As the sample size increases, the sampling distribution of the mean approaches a Normal distribution, regardless of the original data distribution.

Key idea:
Even if data is:

skewed
irregular
non-normal
The sample mean becomes normally distributed for large samples (n ≥ 30).

Relation to Normal Distribution:

CLT is the main reason why the Normal distribution appears everywhere in statistics.

 5. What is skewness in distributions and which distributions exhibit it?

 Skewness refers to the asymmetry of a distribution.

Types:
Positive Skew (Right-skewed):
Tail extends to the right
Example: Poisson distribution (for small λ)

Negative Skew (Left-skewed):
Tail extends to the left

Zero Skew (Symmetric):
Balanced on both sides
Example: Normal distribution

Which distributions show skewness?

Binomial → Can be skewed (depends on p)
Poisson → Usually positively skewed
Normal → No skew (perfectly symmetric)
